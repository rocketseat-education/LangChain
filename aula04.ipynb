```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()
```

```python
model = init_chat_model("openai:gpt-4o-mini")
```

```python
resposta = model.invoke("Explique o que é o LangChain em uma frase.")
print(resposta.text)
```

```python
resposta = model.invoke("Conte uma curiosidade sobre o Brasil.")

print(resposta.text)            # O texto da resposta
print(resposta.content)         # Conteudo bruto(seja: texto, imagem, etc)
print(resposta.usage_metadata)  # Quantos token foram usados
print(resposta.tool_calls)      # Ferramentas que o modelo quis chamar (se houver)
```

```python
from langchain.messages import SystemMessage, HumanMessage
mensagens = [
    SystemMessage(content="Você é um professor que responde de forma simples, para iniciantes."),
    HumanMessage(content="O que é um token em um LLM?")
]
resposta = model.invoke(mensagens)
print(resposta.text)            
```

```python
resposta = model.invoke([
    {"role": "system", "content": "Responda em uma frase."},
    {"role": "user", "content": "O que é LangChain?"}
])

print(resposta.text)
```

```python
# imprime a resposta conforme ela é gerada
for chunk in model.stream("Escreva um parágrafo curto sobre a Amazônia."):
    print(chunk.text, end="", flush=True) #end="" para não pular linha; flush=True para imprimir imediatamente
print() #quebra de linha no final
```

```python
perguntas = [
    "Diga 'oi' em francês.",
    "Diga 'oi' em japonês.",
    "Diga 'oi' em alemão.",
]

respostas = model.batch(perguntas) # -> lista de AIMessage, na mesma ordem

for r in respostas:
    print(r.text)
```
