```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()   #carrega a variável de ambiente GOOGLE_API_KEY do arquivo .env

model = init_chat_model("openai:gpt-4o-mini")

# Cada tupla é (papel, conteúdo). As {chaves} são variáveis dinâmicas.
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um professor de {materia}. Responda de forma didatica."),
    ("user", "Explique o conceito de {conceito} para um iniciante.")
])
```

```python
mensagens = prompt.invoke({
    "materia": "astronomia",
    "conceito": "buraco negro"
})
print(mensagens)
```

```python
mensagens = prompt.invoke({"materia": "astronomia", "conceito": "buraco negro"})
resposta = model.invoke(mensagens)
print(resposta.text)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um professor de {materia}. Responda de forma didatica."),
    ("user", "Explique o conceito de {conceito} para um iniciante.")
])

# Reutilizando MUITAS vezes, com inputs diferentes:
perguntas = [
    {"materia": "biologia", "conceito": "fotossíntese"},
    {"materia": "história", "conceito": "Revolução Industrial"},
    {"materia": "computação", "conceito": "recursão"},
]

for p in perguntas:
    resposta = model.invoke(prompt.invoke(p))
    print(f"--- {p['conceito']} ---")
    print(resposta.text, "\n")
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um assistente de atendimento simpático."),
    MessagesPlaceholder("historico"),   # aqui entram as mensagens anteriores
    ("user", "{pergunta}"),
])

from langchain.messages import HumanMessage, AIMessage

mensagens = prompt.invoke({
    "historico": [
        HumanMessage("Meu pedido chegou quebrado."),
        AIMessage(content="Sinto Muito! Vou abrir uma troca para você."),
    ],
    "pergunta": "Quanto tempo demora a troca?"
})
# O modelo recebe: system + as 2 mensagens do historico + a nova pergunta

for m in mensagens:
    print("#####")
    print(m)
```

```python
from langchain_core.prompts import PromptTemplate
template = PromptTemplate.from_template(
    "Traduza o texto a seguir para {idioma}:\n\n{texto}"
)

texto_final = template.invoke({"idioma": "frances", "texto": "Bom dia!"})
print(texto_final.text) #"Traduza o texto a seguir para frances:\n\nBom dia!"
```

```python
# Versão especilizada: o idioma já vem "travado" em francês
para_frances = template.partial(idioma="frances")

# Agora só falta o texto a cada chamada:
t1 = para_frances.invoke({"texto": "Bom dia!"})
t2 = para_frances.invoke({"texto": "Até logo!"})

print(t1)
print(t2)
```

```python
from langchain_core.prompts import (
    ChatPromptTemplate,
    FewShotChatMessagePromptTemplate,
)

# 1) Exemplos: queremos classificar o sentimento de frases
exemplos = [
    {"frase": "Adorei o produto, chegou rápido!", "rotulo": "positivo"},
    {"frase": "Veio quebrado e o suporte não respondeu", "rotulo": "negativo"},
]

# 2) Molde de UM exemplo (como cada par vira mensagens)
exemplo_prompt = ChatPromptTemplate.from_messages([
    ("user", "{frase}"),
    ("assistant", "{rotulo}"),
])

# 3) Junta os exemplos no formato Few-Shot
few_shot = FewShotChatMessagePromptTemplate(
    examples=exemplos,
    example_prompt=exemplo_prompt,
)

# 4) Prompt final: instrução + exemplos + a nova frase
prompt = ChatPromptTemplate.from_messages([
    ("system", "Classifique o sentimento da frase como 'positivo' ou 'negativo'."),
    few_shot,                   #injeta os exemplos aqui
    ("user", "{frase}"),
])

resposta = model.invoke(prompt.invoke({"frase": "Entrega no prazo e ótimo atendimento."}))
print(resposta.text)  # -> positivo
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "Você é um professor de {materia}. Responda de forma didatica."),
    ("user", "Explique o conceito de {conceito} para um iniciante.")
])

# Em vez de dois invokes separados...
mensagens = prompt.invoke({"materia": "quimica", "conceito": "pH"})
resposta = model.invoke(mensagens)

# ...conecte tudo numa "chain": a saída do prompt vira entrada do modelo
chain = prompt | model
resposta = chain.invoke({"materia": "quimica", "conceito": "pH"})
print(resposta.text)  
```
